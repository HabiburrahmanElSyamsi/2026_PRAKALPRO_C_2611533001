## 2.1 Dasar Teori
Perulangan (looping) adalah struktur kendali yang memungkinkan sekumpulan instruksi dijalankan berulang kali selama syarat tertentu terpenuhi. Dengan perulangan, program tidak perlu menuliskan instruksi yang sama secara manual berkali-kali, sehingga kode menjadi lebih ringkas, mudah dibaca, dan mudah diubah. Dalam Python, terdapat dua bentuk perulangan utama, yaitu for dan while.

Perulangan for digunakan ketika jumlah iterasi sudah diketahui atau dapat ditentukan sebelumnya. Pada setiap iterasi, variabel pencacah (iterator) mengambil satu nilai dari suatu urutan data, lalu blok instruksi di dalamnya dieksekusi. Proses ini berhenti secara otomatis ketika seluruh nilai pada urutan telah dijalani. Urutan yang ditelusuri dapat berupa deret angka dari fungsi range(), string, list, maupun kumpulan data lainnya.

Pada pola dua dimensi seperti tabel, matriks, atau bangun geometri, satu perulangan saja tidak cukup. Dibutuhkan perulangan bersarang (nested loop), yaitu perulangan di dalam perulangan lain. Perulangan luar biasanya mengatur baris, sedangkan perulangan dalam mengatur isi setiap baris, seperti spasi, angka, atau simbol. Setiap kali perulangan luar berjalan satu kali, perulangan dalam dijalankan sampai selesai. Karena itu, ketepatan formula jumlah iterasi pada tiap perulangan sangat menentukan hasil keluaran.

1. Perulangan for
Struktur perulangan dengan jumlah iterasi yang terdefinisi jelas. Pada setiap iterasi, variabel pencacah menerima nilai berikutnya dari urutan data, lalu blok instruksi dijalankan. Perulangan berhenti setelah semua nilai dalam urutan habis diproses.

2. Fungsi range()
Fungsi yang menghasilkan deret bilangan bulat untuk dijadikan acuan iterasi. Fungsi ini dapat menerima satu, dua, atau tiga parameter, yaitu nilai awal, nilai akhir, dan ukuran langkah. Nilai akhir bersifat eksklusif, artinya tidak ikut dihasilkan. Sebagai contoh, range(2, 11, 3) menghasilkan deret 2, 5, dan 8. Dengan langkah bernilai negatif, deret dapat dibuat menurun, misalnya range(9, 0, -2) yang menghasilkan 9, 7, 5, 3, dan 1.

3. Variabel Pencacah (Counter)
Variabel yang menampung nilai saat ini pada setiap putaran perulangan. Nilainya berubah mengikuti deret yang dihasilkan sehingga dapat dipakai dalam perhitungan, misalnya untuk menentukan jumlah karakter atau posisi tertentu pada keluaran program.

4. Perulangan Bersarang (Nested Loop)
Perulangan yang berada di dalam perulangan lain. Perulangan luar biasanya mengatur baris, sedangkan perulangan dalam mengatur isi setiap baris. Jumlah total iterasi adalah hasil kali iterasi perulangan luar dengan iterasi perulangan dalam. Misalnya, perulangan luar sebanyak 5 kali dan perulangan dalam sebanyak 4 kali menghasilkan 20 kali eksekusi blok terdalam.

5. Pencetakan dengan print() dan Parameter end
Secara bawaan, print() menambahkan karakter pindah baris setelah mencetak. Parameter end mengubah perilaku tersebut sehingga beberapa pemanggilan print() dapat tercetak pada baris yang sama. Pemanggilan print() tanpa argumen digunakan untuk berpindah ke baris berikutnya setelah satu baris selesai dibentuk.

6. Masukan Dinamis dengan input() dan int()
Fungsi input() membaca masukan pengguna dalam bentuk string. Agar dapat dipakai dalam perhitungan dan sebagai batas perulangan, nilai tersebut dikonversi ke bilangan bulat menggunakan int(). Dengan cara ini, ukuran pola dapat berubah sesuai masukan tanpa mengubah isi program.

7. Formula Matematis Pembentuk Pola
Pola dua dimensi dibentuk dengan menghubungkan nomor baris dan ukuran pola ke dalam rumus perhitungan. Misalnya, pada pola segitiga rata kanan dengan tinggi n, jumlah spasi di depan setiap baris dapat dinyatakan sebagai n dikurangi nomor baris, sedangkan jumlah bintang adalah dua kali nomor baris dikurangi satu. Rumus seperti ini membuat bentuk pola konsisten untuk berbagai ukuran.

8. Alur Logika dan Penelusuran Program (Tracing)
Alur logika adalah urutan langkah yang dijalankan program dari awal sampai akhir. Penelusuran dilakukan dengan mencatat nilai variabel pada setiap iterasi untuk memastikan keluaran sesuai rencana serta menemukan kesalahan logika, seperti batas range() yang keliru atau jumlah spasi yang kurang.

## 2.2 Kode Program

### 2.2.1 Program jumlah_genap "jumlah_genap_2611533001.py"

```python
ulang_3001 = int(input("Masukkan nilai batas: "))

jumlah_3001 = 0
pertama_3001 = True

for i_3001 in range(1, ulang_3001 + 1):
    if i_3001 % 2 == 0:
        if not pertama_3001:
            print(" + ", end="")
        print(i_3001, end="")
        jumlah_3001 += i_3001
        pertama_3001 = False

print(" =", jumlah_3001)
print("Jumlah =", jumlah_3001)
```

**Penjelasan :**
1. Bagian pertama berfungsi untuk menerima masukan dan menyiapkan variabel. Fungsi input() meminta pengguna mengisi nilai batas, lalu int() mengubahnya menjadi bilangan bulat dan menyimpannya di ulang_3001. Variabel jumlah_3001 diberi nilai awal 0 sebagai penampung hasil penjumlahan. Variabel pertama_3001 bernilai True sebagai penanda bahwa bilangan genap yang akan dicetak adalah yang pertama, sehingga belum perlu diberi tanda + di depannya.

2. Bagian kedua berfungsi untuk menelusuri bilangan dengan perulangan for. Fungsi range(1, ulang_3001 + 1) menghasilkan deret 1 sampai nilai batas, dengan + 1 agar nilai batas ikut diproses. Pada setiap putaran, i_3001 menampung bilangan saat ini. Percabangan if i_3001 % 2 == 0 kemudian menyaring bilangan genap, yaitu bilangan yang sisa baginya dengan 2 sama dengan 0. Bilangan ganjil dilewati.

3. Bagian ketiga berfungsi untuk mencetak bilangan genap dalam bentuk penjumlahan. Jika bilangan genap yang ditemukan bukan yang pertama (not pertama_3001), program mencetak tanda " + " terlebih dahulu. Setelah itu bilangan genap dicetak pada baris yang sama menggunakan end="". Dengan cara ini tanda + hanya muncul di antara bilangan dan tidak ada di awal atau akhir deret.

4. Bagian keempat berfungsi untuk mengakumulasi hasil. Perintah jumlah_3001 += i_3001 menambahkan setiap bilangan genap yang ditemukan ke variabel penampung. Setelah itu, pertama_3001 diubah menjadi False supaya bilangan genap berikutnya diberi tanda + di depannya.

5. Bagian kelima berfungsi untuk menampilkan hasil akhir. Setelah perulangan selesai, print(" =", jumlah_3001) melengkapi baris penjumlahan dengan tanda sama dengan dan totalnya. Pindah baris terjadi otomatis karena print() ini tidak memakai end="". Baris terakhir, print("Jumlah =", jumlah_3001), menampilkan total sekali lagi dalam kalimat tersendiri.

### 2.2.2 Program nested_for1 "nested_for1_2611533001.py"

```python
batas_3001 = int(input("Masukkan nilai batas: "))
for line_3001 in range(1, batas_3001 + 1):
    for j_3001 in range(1, (-1 * line_3001 + batas_3001) + 1):
        print(".", end="")
    print(line_3001)
```

**Penjelasan :**
1. Bagian pertama berfungsi untuk menerima masukan. Fungsi input() meminta pengguna mengisi nilai batas, lalu int() mengubahnya menjadi bilangan bulat dan menyimpannya di batas_3001. Nilai ini menentukan jumlah baris yang dicetak sekaligus panjang titik pada baris pertama.

2. Bagian kedua berfungsi sebagai perulangan luar yang mengatur baris. Fungsi range(1, batas_3001 + 1) menghasilkan deret 1 sampai nilai batas, dengan + 1 agar nilai batas ikut diproses. Pada setiap putaran, line_3001 menampung nomor baris saat ini. Satu putaran perulangan luar sama dengan satu baris keluaran.

3. Bagian ketiga berfungsi sebagai perulangan dalam yang mencetak titik. Batas atas range() adalah (-1 * line_3001 + batas_3001) + 1, yang sama dengan batas_3001 - line_3001 + 1. Artinya j_3001 berjalan dari 1 sampai batas_3001 - line_3001, sehingga titik yang tercetak sebanyak batas_3001 - line_3001. Semakin besar nomor baris, semakin sedikit titiknya. Pada baris terakhir, jumlah titiknya 0 sehingga perulangan dalam tidak berjalan sama sekali. Parameter end="" membuat semua titik tercetak pada baris yang sama.

4. Bagian keempat berfungsi untuk mencetak nomor baris dan pindah baris. Setelah perulangan dalam selesai, print(line_3001) mencetak nomor baris tepat setelah deretan titik. Karena tidak memakai end="", print() ini sekaligus memindahkan kursor ke baris berikutnya. Perintah ini sejajar dengan for j_3001, sehingga hanya dijalankan sekali per baris, bukan setiap kali sebuah titik dicetak.

5. Secara keseluruhan, pola yang terbentuk adalah deret angka yang rata kanan. Titik berfungsi sebagai penyeimbang di sebelah kiri, dengan rumus jumlah titik = batas - nomor baris. Konsep ini sama dengan spasi penyeimbang 2 * (N - baris) pada program Jam Pasir.

Pada setiap putaran, i_3001 menampung nomor baris saat ini. Satu putaran perulangan luar sama dengan satu baris keluaran, sehingga jumlah baris yang tercetak sama dengan batas_3001.

Karena itu, j_3001 selalu berjalan sebanyak batas_3001 kali pada setiap barisnya. Jumlah total bintang adalah hasil kali iterasi perulangan luar dan perulangan dalam, yaitu batas_3001 × batas_3001. Hal ini sesuai konsep perulangan bersarang, yaitu perulangan luar mengatur baris dan perulangan dalam mengatur isi setiap baris.

Nilai terkecilnya 0 (pojok kiri atas) dan nilai terbesarnya 2 × batas (pojok kanan bawah). Jumlah total angka yang dicetak adalah (batas_3001 + 1) × (batas_3001 + 1).

### 2.2.5 Program nested_for4 "nested_for4_2611533001.py"

```python
tinggi_3001 = int(input("Masukkan tinggi pola (bilangan genap, misal 10): "))

if tinggi_3001 % 2 != 0:
    print("Tinggi harus bilangan genap!")
else:
    a_3001 = tinggi_3001
    c_3001 = a_3001
    lebar_3001 = (2 * tinggi_3001) - 2

    for i_3001 in range(1, tinggi_3001 + 1):
        b_3001 = c_3001 + 1

        for j_3001 in range(1, lebar_3001 + 1):

            if i_3001 == 1 or i_3001 == tinggi_3001:
                if j_3001 == 1 or j_3001 == lebar_3001:
                    print("#", end="")
                else:
                    print("=", end="")
            else:
                if j_3001 == 1 or j_3001 == lebar_3001:
                    print("|", end="")
                else:
                    if j_3001 == c_3001:
                        print("<", end="")
                    elif j_3001 == b_3001:
                        print(">", end="")
                    elif j_3001 == (lebar_3001 - c_3001):
                        print("<", end="")
                    elif j_3001 == (lebar_3001 - c_3001 + 1):
                        print(">", end="")
                    elif j_3001 > b_3001 and j_3001 < (lebar_3001 - c_3001):
                        print(".", end="")
                    else:
                        print(" ", end="")

        print()                      # <-- pindah baris, di dalam loop i

        a_3001 -= 2                  # <-- update juga di dalam loop i
        if a_3001 <= 0:
            c_3001 = (-a_3001) + 2
        else:
            c_3001 = a_3001
```

**Penjelesan:**
1. Bagian pertama berfungsi untuk menerima masukan dan memvalidasinya. Fungsi input() meminta pengguna mengisi tinggi pola, lalu int() mengubahnya menjadi bilangan bulat dan menyimpannya di tinggi_3001. Percabangan if tinggi_3001 % 2 != 0 memeriksa apakah tinggi ganjil. Jika ya, program hanya mencetak pesan kesalahan. Jika genap, blok else yang berisi seluruh proses pencetakan pola dijalankan.

2. Bagian kedua berfungsi untuk menyiapkan variabel pengatur pola. Variabel a_3001 dan c_3001 diberi nilai awal sebesar tinggi_3001. Variabel c_3001 adalah posisi kolom tempat tanda < kiri dicetak, sedangkan a_3001 adalah penghitung bantu untuk memperbarui c_3001 di setiap baris. Variabel lebar_3001 dihitung dengan rumus (2 * tinggi_3001) - 2, sehingga untuk tinggi 10 lebar polanya 18 kolom.

3. Bagian ketiga berfungsi sebagai perulangan luar dan dalam. Perulangan luar range(1, tinggi_3001 + 1) mengatur baris, dengan i_3001 sebagai nomor baris. Pada awal setiap baris, b_3001 = c_3001 + 1 dihitung sebagai posisi tanda > yang berpasangan dengan < di sebelahnya. Perulangan dalam range(1, lebar_3001 + 1) mengatur kolom, dengan j_3001 sebagai nomor kolom. Perulangan dalam ini menentukan satu karakter untuk setiap kolom, dan semua karakter tercetak pada baris yang sama karena end="".

4. Bagian keempat berfungsi untuk mencetak bingkai. Jika i_3001 adalah baris pertama atau terakhir, pojok kiri dan kanan (j_3001 == 1 atau j_3001 == lebar_3001) dicetak #, sedangkan kolom di antaranya dicetak =. Pada baris lainnya, kolom pertama dan terakhir dicetak | sebagai dinding tegak.

5. Bagian kelima berfungsi untuk mencetak isi pola di dalam dinding. Setiap kolom diperiksa secara berurutan dengan if-elif:

### 2.2.8 Program perulangan_for3 "perulangan_for3_2611533001.py"

```python
ulang_3001 = int(input("Masukkan jumlah perulangan: "))

jumlah_3001 = 0
for i_3001 in range(1, ulang_3001 + 1):
    print(i_3001, end=" ")
    jumlah_3001 = jumlah_3001 + i_3001

    if i_3001 < ulang_3001:
        print(" + ", end=" ")
    else:
        print(" = ", jumlah_3001, end=" ")
print()
print("Jumlah =", jumlah_3001)
```

**Penjelesan:**
1. Bagian pertama berfungsi untuk menerima masukan dan menyiapkan variabel. Fungsi input() meminta pengguna mengisi jumlah perulangan, lalu int() mengubahnya menjadi bilangan bulat dan menyimpannya di ulang_3001. Variabel jumlah_3001 diberi nilai awal 0 sebagai penampung hasil penjumlahan.

2. Bagian kedua berfungsi sebagai perulangan for yang menelusuri bilangan. Fungsi range(1, ulang_3001 + 1) menghasilkan deret 1 sampai ulang_3001, dengan + 1 agar nilai batas ikut diproses. Pada setiap putaran, i_3001 menampung bilangan saat ini. Bilangan tersebut dicetak dengan print(i_3001, end=" ") pada baris yang sama, lalu langsung ditambahkan ke penampung melalui jumlah_3001 = jumlah_3001 + i_3001.

3. Bagian ketiga berfungsi untuk mencetak tanda + di antara bilangan. Percabangan if i_3001 < ulang_3001 bernilai benar selama bilangan saat ini belum mencapai bilangan terakhir. Pada kondisi ini, program mencetak " + " dengan end=" " sehingga baris tidak berpindah dan bilangan berikutnya tercetak di sebelahnya. Dengan cara ini, tanda + hanya muncul di antara bilangan dan tidak ada setelah bilangan terakhir.

4. Bagian keempat berfungsi untuk mencetak hasil pada bilangan terakhir. Blok else dijalankan ketika i_3001 sama dengan ulang_3001. Perintah print(" = ", jumlah_3001, end=" ") mencetak tanda sama dengan diikuti total penjumlahan, sehingga deret penjumlahan ditutup dengan hasilnya pada baris yang sama.

5. Bagian kelima berfungsi untuk pindah baris dan menampilkan hasil akhir. Perintah print() tanpa argumen memindahkan kursor ke baris berikutnya setelah perulangan selesai. Perintah ini sejajar dengan for, sehingga hanya dijalankan sekali. Baris terakhir, print("Jumlah =", jumlah_3001), menampilkan total sekali lagi dalam kalimat tersendiri.

### 2.2.9 Program latihan5 "latihan5_2611533001.py"

```python
tinggi_3001 = int(input("Masukkan tinggi segitiga: "))

for i_3001 in range(1, tinggi_3001 + 1):
    print(" " * (tinggi_3001 - i_3001), end="")
    for j_3001 in range(1, i_3001 + 1):
        print("* ", end="")
    print()
```

Semakin besar nomor baris, semakin sedikit spasinya, sehingga baris bawah bergeser ke kiri dan bentuk segitiga terbentuk. Parameter end="" menjaga agar bintang berikutnya tercetak pada baris yang sama. Perlu dicatat, program ini memakai perkalian string, sedangkan soal Jam Pasir melarang cara tersebut dan mewajibkan perulangan for bersarang murni.

```python
print("=== PROGRAM JAM PASIR KRISTAL PALINDROMIK (PEKAN 5) ===")
n_3001 = int(input("Masukkan ukuran skala jam pasir (N): "))

# ---------- Bingkai atas ----------
print("#", end="")
for garis_3001 in range(4 * n_3001 + 5):
    print("=", end="")
print("#")

# ---------- Fase 1: jam pasir atas (baris N turun s.d. 1) ----------
for baris_3001 in range(n_3001, 0, -1):
    print("|", end="")
    print(" ", end="")
    # spasi penyeimbang kiri
    for spasi_kiri_3001 in range(2 * (n_3001 - baris_3001)):
        print(" ", end="")
    # deret angka mundur (baris -> 1)
    for angka_mundur_3001 in range(baris_3001, 0, -1):
        print(angka_mundur_3001, end=" ")
    # poros kristal
    print("<*>", end="")
    # deret angka maju (1 -> baris)
    for angka_maju_3001 in range(1, baris_3001 + 1):
        print(" ", end="")
        print(angka_maju_3001, end="")
    # spasi penyeimbang kanan
    for spasi_kanan_3001 in range(2 * (n_3001 - baris_3001)):
        print(" ", end="")
    print(" ", end="")
    print("|", end="")
    print()

# ---------- Fase 2: poros titik pusat ----------
print("|", end="")
for spasi_kiri_3001 in range(2 * n_3001 + 1):
    print(" ", end="")
print("<*>", end="")
for spasi_kanan_3001 in range(2 * n_3001 + 1):
    print(" ", end="")
print("|", end="")
print()

# ---------- Fase 3: jam pasir bawah (baris 1 naik s.d. N) ----------
for baris_3001 in range(1, n_3001 + 1):
    print("|", end="")
    print(" ", end="")
    # spasi penyeimbang kiri
    for spasi_kiri_3001 in range(2 * (n_3001 - baris_3001)):
        print(" ", end="")
    # deret angka mundur (baris -> 1)
    for angka_mundur_3001 in range(baris_3001, 0, -1):
        print(angka_mundur_3001, end=" ")
    # poros kristal
    print("<*>", end="")
    # deret angka maju (1 -> baris)
    for angka_maju_3001 in range(1, baris_3001 + 1):
        print(" ", end="")
        print(angka_maju_3001, end="")
    # spasi penyeimbang kanan
    for spasi_kanan_3001 in range(2 * (n_3001 - baris_3001)):
        print(" ", end="")
    print(" ", end="")
    print("|", end="")
    print()

# ---------- Bingkai bawah ----------
print("#", end="")
for garis_3001 in range(4 * n_3001 + 5):
    print("=", end="")
print("#")
```

Bingkai tidak memakai perkalian string, sehingga setiap karakter = lahir dari satu iterasi.
