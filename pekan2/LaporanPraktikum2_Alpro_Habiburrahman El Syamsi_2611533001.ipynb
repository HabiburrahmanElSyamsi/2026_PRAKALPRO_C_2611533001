## 1.1 Latar Belakang
Dalam dunia pemrograman dan ilmu komputer, pemrosesan data merupakan fungsi utama dari setiap perangkat lunak. Untuk dapat mengolah data dengan benar dan efisien, sebuah program harus memahami karakteristik dari data yang diproses. Oleh karena itu, konsep tipe data dan variabel menjadi fondasi paling mendasar yang wajib dikuasai oleh setiap pemrogram sebelum mempelajari struktur data yang lebih kompleks.

Variabel berperan sebagai wadah atau tempat penyimpanan sementara di dalam memori komputer yang digunakan untuk menampung nilai selama program dijalankan. Agar data dalam variabel dapat diakses dan diolah secara optimal, komputer memerlukan aturan penamaan serta kejelasan mengenai jenis nilai yang disimpan, yang ditentukan melalui tipe data. Selain variabel dinamis, pengembangan program juga sering memerlukan nilai tetap yang tidak boleh diubah sepanjang eksekusi program, yang dikenal sebagai konstanta. Penggunaan konstanta bersama dengan penanganan masukan (input) dari pengguna memungkinkan program berjalan secara dinamis namun tetap terkontrol.

## 2.1 Dasar Teori

1. Variabel dan Memori
Variabel adalah tempat menyimpan nilai sementara di memori komputer selama program berjalan. Elemen utamanya terdiri dari nama variabel (identitas) dan nilai (data yang tersimpan dan dapat diubah).
2. Tipe Data
Tipe data menentukan jenis nilai yang disimpan agar komputer tahu cara mengolahnya. Tipe data dasar meliputi:

int: Bilangan bulat (contoh: 10).

float: Bilangan desimal (contoh: 3.14).

complex: Bilangan kompleks dengan bagian riil dan imajiner (contoh: 2 + 3j).

string: Teks atau kumpulan karakter (contoh: "Data").

boolean: Nilai logika kebenaran (True atau False).

3. Konstanta
Konstanta adalah variabel yang nilainya bernilai tetap dan tidak boleh diubah selama program berjalan (contoh: nilai $\pi = 3.14$).
4. Operasi Input dan Output
Mekanisme interaksi program: Input menerima masukan data dari pengguna melalui keyboard, sedangkan Output menampilkan hasil pemrosesan ke layar. Data input teks biasanya perlu dikonversi (type casting) jika ingin diolah sebagai angka.

### 2.2.3 Program Driver "Konstanta_2611533001"

```python
from typing import Final
PI: Final = 3.14
print("Pi: %f" % (PI))
jari_3001 = float(input("Masukkan nilai jari-jari: "))
luas_3001 = PI * jari_3001
print("Luas lingkaran dengan jari-jari %.2f adalah %.2f" % (jari_3001,luas_3001))
```

Penjelasan

1. Pemanggilan Final dari library typing from typing import Final
2. Deklarasi PI: Final = 3.14 sebagai konstanta
3. Cetak ("Pi: %f" % (PI))
4. Nilai jari_3001 = float(input("Masukkan nilai jari-jari: "))
5. Nilai luas_3001 = PI * jari_3001
6. Cetak ("Luas lingkaran dengan jari-jari %.2f adalah %.2f" % (jari_3001,luas_3001))
