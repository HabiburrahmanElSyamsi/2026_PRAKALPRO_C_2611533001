## 2.1 Dasar Teori
Operator merupakan simbol khusus dalam bahasa pemrograman Python yang digunakan untuk melakukan operasi tertentu pada satu atau lebih nilai yang disebut sebagai operand. Sebagai contoh, dalam ekspresi a + b, simbol + bertindak sebagai operator, sedangkan a dan b adalah operand. Operator menjadi instrumen utama dalam manipulasi data, perhitungan matematis, serta pembuatan logika pengkondisian pada program.

1. Operator Aritmatika (Arithmetic Operators)
Operator aritmatika digunakan untuk melakukan operasi matematika dasar pada tipe data numerik (seperti integer dan float).
Penjumlahan (+): Menambahkan dua operand.
Pengurangan (-): Mengurangi operand kanan dari operand kiri.
Perkalian (*): Mengalikan dua operand.
Pembagian (/): Membagi operand dan menghasilkan nilai bertipe float.
Pembagian Bulat / Floor Division (//): Membagi operand dan membulatkan hasilnya ke bawah ke bilangan bulat terdekat.
Sisa Bagi / Modulus (%): Mengembalikan sisa hasil dari pembagian dua operand.
Pangkat / Exponentiation (**): Memangkatkan operand kiri dengan operand kanan.

2. Operator Perbandingan (Comparison/Relational Operators)
Operator perbandingan digunakan untuk membandingkan dua buah nilai. Hasil dari operasi perbandingan selalu menghasilkan nilai boolean, yaitu True (benar) atau False (salah).
Sama dengan (==): Bernilai True jika kedua operand bernilai sama.
Tidak sama dengan (!=): Bernilai True jika kedua operand tidak bernilai sama.
Lebih besar dari (>): Bernilai True jika operand kiri lebih besar dari operand kanan.
Lebih kecil dari (<): Bernilai True jika operand kiri lebih kecil dari operand kanan.
Lebih besar atau sama dengan (>=): Bernilai True jika operand kiri lebih besar dari atau sama dengan operand kanan.
Lebih kecil atau sama dengan (<=): Bernilai True jika operand kiri lebih kecil dari atau sama dengan operand kanan.

3. Operator Logika (Logical Operators)
Operator logika digunakan untuk menggabungkan atau mengevaluasi beberapa kondisi boolean.
and: Mengembalikan nilai True hanya jika kedua kondisi/operand bernilai True.
or: Mengembalikan nilai True jika salah satu atau kedua kondisi/operand bernilai True.
not: Membalikkan status boolean operand (jika True menjadi False, dan sebaliknya).

4. Operator Penugasan (Assignment Operators)
Operator penugasan digunakan untuk memberikan atau memperbarui nilai pada sebuah variabel. Selain operator penugasan standar (=), Python menyediakan compound assignment operators yang menggabungkan operasi aritmatika dengan penugasan secara langsung.
Penugasan Dasar (=): Memberikan nilai di ruas kanan ke variabel di ruas kiri (contoh: x = 5).
Penambahan dan Penugasan (+=): Menambahkan nilai dan memperbarui variabel (contoh: x += 3 identik dengan x = x + 3).
Pengurangan dan Penugasan (-=): Mengurangi nilai dan memperbarui variabel (contoh: x -= 2 identik dengan x = x - 2).
Perkalian dan Pembagian Penugasan (*=, /=, //=, %=, **=): Melakukan operasi aritmatika terkait diikuti dengan pembaruan nilai variabel.

5. Operator Keanggotaan (Membership Operators)
Operator keanggotaan digunakan untuk menguji apakah suatu nilai atau elemen terdapat di dalam suatu urutan/koleksi data (seperti string, list, tuple, atau dictionary).
in: Bernilai True jika nilai yang dicari ditemukan di dalam sekuens/koleksi.
not in: Bernilai True jika nilai yang dicari tidak ditemukan di dalam sekuens/koleksi.

6. Operator Identitas (Identity Operators)
Operator identitas digunakan untuk membandingkan lokasi memori (alokasi objek) dari dua buah variabel, bukan sekadar membandingkan nilainya saja.
is: Bernilai True jika kedua variabel menunjuk ke objek yang sama di memori.
is not: Bernilai True jika kedua variabel menunjuk ke objek yang berbeda di memori.

7. Operator Bitwise (Bitwise Operators)
Operator bitwise mengoperasikan nilai pada tingkat biner (bit demi bit). Operator ini biasanya digunakan pada operasi komputasi tingkat rendah. Contoh dari operator bitwise meliputi AND (&), OR (|), XOR (^), NOT (~), Left Shift (<<), dan Right Shift (>>).

### 2.2.2 Program "assigment_2611533001"

```python
angka1_3001 = int(input("Input angka-1: "))
angka2_3001 = int (input("Input angka-2: "))

print("\nNilai awal angka1_3001 =", angka1_3001)
print("Nilai angka2_3001 =", angka2_3001)

# Assignment biasa
hasil = angka1_3001
print("\nAssignment biasa (=)")
print("Hasil =", hasil)

# Assignment penambahan
hasil = angka1_3001
hasil += angka2_3001
print("\nAssignment penambahan (+=)")
print("Hasil =", hasil)

# Assignment pengurangan
hasil = angka1_3001
hasil -= angka2_3001
print("\nAssignment pengurangan (-=)")
print("Hasil =", hasil)

# Assignment perkalian
hasil = angka1_3001
hasil *= angka2_3001
print("\nAssignment perkalian (*=)")
print("Hasil =", hasil)

# Assignment pembagian, pembagian bulat, dan sisa bagi

if angka2_3001 != 0:
    hasil = angka1_3001
    hasil /= angka2_3001
    print("\nAssignment pembagian (/=)")
    print("Hasil =", hasil)
    # Operator tambahan
    hasil = angka1_3001
    hasil //= angka2_3001
    print("\nAssignment pembagian bulat (//=)")
    print("Hasil =", hasil)
    hasil = angka1_3001
    hasil %= angka2_3001
    print("\nAssignment sisa bagi (%=)")
    print("Hasil =", hasil)
else:
    print("\nPembagian tidak dapat dilakukan.")
    print("Angka kedua tidak boleh bernilai 0.")

# Operator tambahan: assignment perpangkatan
hasil = angka1_3001
hasil **= angka2_3001
print("\nAssignment perpangkatan (**=)")
print("Hasil =", hasil)

# Bitwise XOR
hasil = angka1_3001 ^ angka2_3001
print("\nBitwise XOR (^)")
print(angka1_3001, "^", angka2_3001, "=", hasil)
print("Biner hasil =", bin(hasil))
print("Biner hasil (8 bit) =", format(hasil, "08b"))

# Bitwise NOT
hasil = ~angka1_3001
print("\nBitwise NOT (~)")
print("~", angka1_3001, "=", hasil)
print("Biner hasil =", bin(hasil))
print("Biner hasil (8 bit) =", format(hasil, "08b"))

# Bitwise geser kiri
jumlah_geser = int(input("\nMasukkan jumlah pergeseran bit: "))

hasil = angka1_3001 << jumlah_geser
print("\nBitwise geser kiri (<<)")
print(angka1_3001, "<<", jumlah_geser, "=", hasil)
print("Biner hasil =", bin(hasil))
print("Biner hasil (8 bit) =", format(hasil, "08b"))

# Bitwise geser kanan
hasil = angka1_3001 >> jumlah_geser
print("\nBitwise geser kanan (>>)")
print(angka1_3001, ">>", jumlah_geser, "=", hasil)
print("Biner hasil =", bin(hasil))
print("Biner hasil (8 bit) =", format(hasil, "08b"))
```

(syarat_jumlah_3001) << 2,"04b")} | {format(int(promo_valid_3001) << 3,"04b")}") Menampilkan masing-masing kondisi dalam bentuk biner 4 bit.

format(nilai, "04b")
digunakan untuk mengubah nilai menjadi biner dengan panjang minimal 4 bit.

Berdasarkan praktikum yang telah dilaksanakan, dapat disimpulkan bahwa operator pada Python merupakan fondasi utama dalam pemrosesan data dan pembentukan logika program.
