## 2.1 Dasar Teori
Tuliskan penjelasan konsep/teori yang mendasari praktikum ini (definisi, elemen-elemen penting, dsb). Bisa memakai poin bernomor untuk merinci sub-konsep, contoh:

1. Konsep Dasar Pengambilan Keputusan (Decision Making):
Conditional statement atau struktur percabangan adalah mekanisme kontrol alur program yang memungkinkan komputer mengeksekusi blok kode tertentu berdasarkan evaluasi kondisi boolean (True atau False). Dalam Python, ekspresi logika ini dibentuk menggunakan operator perbandingan (==, !=, >, <, >=, <=) dan operator logika (and, or, not).

2. Struktur Sintaks if, elif, dan else:
Python menyediakan tiga komponen utama untuk menangani logika kondisional secara fleksibel:

-Pernyataan if: Blok dasar yang akan dieksekusi jika kondisi utama bernilai True.
-Pernyataan elif (Else If): Digunakan untuk menguji kondisi alternatif secara berurutan apabila kondisi if sebelumnya bernilai False.
-Pernyataan else: Blok penampung akhir yang dieksekusi jika seluruh kondisi if dan elif bernilai False.

3. Aturan Indentasi dan Nested Conditional (Percabangan Bersarang):
Python memanfaatkan aturan indentasi (spasi atau tab) sebagai penanda blok kode kondisional pengganti tanda kurung kurawal {}. Selain itu, Python mendukung nested conditional, yaitu struktur di dalam blok if atau else terdapat pernyataan kondisional lainnya untuk menangani logika pengambilan keputusan bertingkat yang lebih kompleks.

### 2.2.5 Program Multi-IF 2 "multi_if2_2611533001.py"

```python
# Input dari user
total_belanja_3001 = float(input("Masukkan total belanja (Rp): "))

# Input status member (mengecek apakh user mengetik 'y' atau 'ya')
input_member_3001 = input("Apakah Anda Member? (y/t): ").strip().lower()
is_member_3001 = input_member_3001 in ["y", "ya"]

# Input status kode promo (mengecek apakah user mengetik 'y' atau 'ya')
input_promo_3001 = input("Apakah kode promo valid? (y/t): ").strip().lower()
kode_promo_valid_3001 = input_promo_3001 in ["y", "ya"]

total_diskon_persen_3001 = 0

if total_belanja_3001 > 1000000:
    total_diskon_persen_3001 += 10 # Diskon belanja besar

if is_member_3001:
    total_diskon_persen_3001 += 5 # Diskon member

if kode_promo_valid_3001:
    total_diskon_persen_3001 += 15 # Diskon voucher

# Menghitung nominal diskon dan total bayar
nominal_diskon_3001 = total_belanja_3001 * (total_diskon_persen_3001 / 100)
total_bayar_3001 = total_belanja_3001 - nominal_diskon_3001

# Output hasil
print("\n--- Rincian Pembayaran ---")
print(f"Total Diskon : {total_diskon_persen_3001}% (Rp {nominal_diskon_3001 :,.0f})")
print(f"Total Bayar : Rp {total_bayar_3001 :,.0f}")

print(f"Total diskon yang anda dapatkan: {total_diskon_persen_3001}")
```

### 2.2.7 Program Match Case 1 "match_case1_2611533001.py"

```python
bulan = int(input("Masukkan angka bulan (1-12): "))

match bulan:
    case 1:
        print("Januari")
    case 2:
        print("Februari")
    case 3:
        print("Maret")
    case 4:
        print("April")
    case 5:
        print("Mei")
    case 6:
        print("Juni")
    case 7:
        print("Juli")
    case 8:
        print("Agustus")
    case 9:
        print("September")
    case 10:
        print("Oktober")
    case 11:
        print("November")
    case 12:
        print("Desember")
    case _:
        print("Angka tidak valid")
```

**Penjelasan :**

1. Pemadanan Pola dengan Match-Case
match bulan_3001:
Mencocokkan nilai variabel bulan_3001 dengan deretan opsi case.

2. Wildcard Case Default
case _:
Simbol _ bertindak sebagai penampung default jika input angka pengguna berada di luar rentang 1–12.

Empat if terpisah (bukan elif) supaya semua kondisi yang terpenuhi ikut menambah persen diskon — inilah maksud "akumulasi/ditumpuk". Kalau dipakai elif, hanya satu diskon yang berlaku, padahal soal minta bisa gabungan.

f-string dengan format :,.0f menampilkan angka dengan pemisah ribuan (koma) dan tanpa desimal, sesuai contoh output di soal (mis. Rp 200,000).
